## **Median Filter**

In this exercise, you will implement a median filter, a non-linear filter that reduces noise in an input image by replacing each pixel with the median value of its surrounding neighborhood. Follow these steps to complete the implementation of the `median_filter()` function:

1. Define the filter size.

2. Add padding to the input image (if padding > 0).

3. Initialize an output matrix based on image size, kernel size, padding and stride.

4. Extract the region for each pixel.

5. Find the median of the region.

6. Set the median value for the corresponding pixel in the output matrix.

7. Return the filtered image.

```python
# Import libraries
import numpy as np
import cv2
import matplotlib.pyplot as plt 

# Load an image
image = cv2.imread('lenna.png', cv2.IMREAD_COLOR)
grey = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Set a kernel
kernel3 = (3, 3)
kernel5 = (5, 5)
kernel7 = (7, 7)

kernel_size = kernel3

# Set needed variables
padding = (kernel_size[0] // 2, kernel_size[1] // 2)
stride = (1, 1)

def median_filter(image, kernel_size, padding, stride):
    """
    Performs a median filtering operation on a multi-channel image using a given kernel.

    Parameters:
      image: A 2D array representing the input image (height, width).
      kernel_size: A tupple representing the kernel size (height, width).
      padding: An tupple representing the amount of padding to be added (vertical, horizontal).
      stride: An tupple representing the step size for moving the kernel (vertical, horizontal).

    Returns:
      An array representing the output image after applying the filtering operation.
    """
    # define filter size
    kH = kernel_size[0]
    kW = kernel_size[1]
    
    # add padding to input image
    pH = padding[0]
    pw = padding[1]
    constant = 0 # not used in reflect mode
    padded_in = np.pad(image, pad_width = padding, mode = 'reflect', reflect_type = 'even')

    # initialize ouput matrix based on parameters
    iH = image.shape[0]
    iW = image.shape[1]
    sH = stride[0]
    sW = stride[1]
    outH = int((iH + 2 * pH - kH) / sH) + 1
    outW = int((iW + 2 * pw - kW) / sW) + 1
    out_sha = (outH, outW)
    filtered_image = np.zeros(out_sha, dtype = np.float32)

    # extract region for each pixel (nested loops)
    for y in range (0, padded_in.shape[0] - kH + 1, sH):
        for x in range(0, padded_in.shape[1] - kW + 1, sW):
            region = padded_in[y:y + kH, x: x + kW]
            # find median value of region
            values = []
            for h in range(kH):
                for w in range(kW):
                    values.append(region[h, w])
            pixel = np.median(values)
            # set median value to output matrix
            filtered_image[y // sH, x // sW] = pixel
    # return output matrix (filtered_image)
    return filtered_image


# Apply your median filter
filtered_image = median_filter(grey, kernel_size, padding, stride)

# Display the image and filtered_image
plt.figure(figsize = (10,5))

# original
orForDisplay = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.subplot(1, 2, 1)
plt.imshow(orForDisplay)
plt.title('Original Image')
plt.axis('off')

# median filter applied
forDisplay = np.clip(filtered_image, 0, 255).astype(np.uint8)
plt.subplot(1,2,2)
plt.imshow(forDisplay, cmap = 'gray')
plt.title('Filtered Image')
plt.axis('off')

#display
plt.tight_layout()
plt.show()
```

After implementing your median filter, test it with different kernel sizes on a noisy image, which can be generated using the provided `salt_pepper()` function.

```python
def salt_pepper(image, salt_prob, pepper_prob):
    """
    Add salt and pepper noise to an image.

    :param image: Input image (NumPy array).
    :param salt_prob: Probability of adding salt noise (white pixels).
    :param pepper_prob: Probability of adding pepper noise (black pixels).
    :return: Noisy image with salt and pepper noise added.
    """
    noisy_image = np.copy(image)

    # Salt noise (white pixels)
    num_salt = np.ceil(salt_prob * image.size).astype(int)
    coords_salt = [np.random.randint(0, i - 1, num_salt) for i in image.shape]
    noisy_image[tuple(coords_salt)] = 255  # For salt (white)

    # Pepper noise (black pixels)
    num_pepper = np.ceil(pepper_prob * image.size).astype(int)
    coords_pepper = [np.random.randint(0, i - 1, num_pepper) for i in image.shape]
    noisy_image[tuple(coords_pepper)] = 0  # For pepper (black)

    return noisy_image
```

```python
# Load an image using OpenCV
image = cv2.imread('cameraman.png', cv2.IMREAD_COLOR)
# Add salt-and-pepper noise
salt_prob = .1
pepper_prob = .1
noisy_image = salt_pepper(image, salt_prob, pepper_prob)

# Display the original and noisy images

plt.figure(figsize = (12,4))

# original
orForDisplay = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.subplot(1, 3, 1)
plt.imshow(orForDisplay)
plt.title('Original Image')
plt.axis('off')

# salt-and-pepper applied
clipped = np.clip(noisy_image, 0, 255).astype(np.uint8)
forDisplay = cv2.cvtColor(clipped, cv2.COLOR_BGR2GRAY)
plt.subplot(1,3,2)
plt.imshow(forDisplay, cmap = 'gray')
plt.title('Salt-and-Pepper Image')
plt.axis('off')

# median filter applied
filtered_noisy_image = median_filter(cv2.cvtColor(noisy_image, cv2.COLOR_BGR2GRAY), kernel_size, padding, stride)
forDisplayFiltered = np.clip(filtered_noisy_image, 0, 255).astype(np.uint8)
plt.subplot(1,3,3)
plt.imshow(forDisplayFiltered, cmap = 'gray')
plt.title('Filtered Noisy Image')
plt.axis('off')

#display
plt.tight_layout()
plt.show()
```

(A). In this exercise, you will implement a **Dilation** function. Dilation is a morphological operation that expands the boundaries of objects in a binary image.

```python
import matplotlib.pyplot as plt, os
import cv2
import numpy as np

def dilation(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    ph, pw = kh//2, kw//2
    pad = np.pad(image, ((ph,ph),(pw,pw)), mode='constant', constant_values=0)
    out = np.zeros_like(image, dtype=np.uint8)
    for i in range(h):
        for j in range(w):
            region = pad[i:i+kh, j:j+kw]
            if np.any(region[kernel==1] == 255):
                out[i,j] = 255
    return out

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))

dilation_image = dilation(FG_MASK, kernel)   

plt.figure(figsize=(8,3))
plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(cv2.imread(SELECTED_IMG_PATH), cv2.COLOR_BGR2RGB))
plt.title(f"Original\n({os.path.basename(SELECTED_IMG_PATH)})")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(FG_MASK, cmap="gray")
plt.title("Binary foreground (white)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(dilation_image, cmap="gray")   
plt.title("B) Dilation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

(B). In this exercise, you will implement an **Erosion** function. Erosion is a morphological operation that shrinks the boundaries of objects in a binary image by removing pixels on object boundaries.

```python
print("Using:", os.path.basename(SELECTED_IMG_PATH))

fg  = FG_MASK.copy()
rgb = RGB_ORIG.copy()

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))

def erosion(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    ph, pw = kh//2, kw//2
    padded = np.pad(image, ((ph, ph), (pw, pw)), mode='constant', constant_values=0)
    out = np.zeros_like(image, dtype=np.uint8)
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            if np.all(region[kernel==1] == 255):
                out[i, j] = 255
    return out

eroded_once = erosion(FG_MASK, kernel)
eroded_twice = erosion(eroded_once, kernel)

eroded_image = erosion(fg, kernel)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(fg, cmap="gray")
plt.title("Original (foreground white)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(eroded_once, cmap="gray")
plt.title("Eroded Once")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(eroded_twice, cmap="gray")
plt.title("Eroded Twice")
plt.axis("off")

plt.tight_layout()
plt.show()
```

(C). In this exercise, you will implement an **Opening** function. Opening is a morphological operation that removes small objects from the foreground of a binary image while preserving the shape and size of larger objects. This is achieved by first applying erosion and then dilation using your previously implemented functions.



```python
print("Using:", os.path.basename(SELECTED_IMG_PATH))

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))

def opening(image, kernel):
    return dilation(erosion(image, kernel), kernel)

opened = opening(FG_MASK, kernel)

import matplotlib.pyplot as plt
plt.figure(figsize=(10,4))
plt.subplot(1,3,1); plt.imshow(RGB_ORIG); plt.title("Original"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(FG_MASK, cmap="gray"); plt.title("Binary mask"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(opened, cmap="gray"); plt.title("C) Opening"); plt.axis("off")
plt.tight_layout(); plt.show()
```

(D) In this exercise, you will implement a **Closing** function. Closing is a morphological operation that fills small holes and gaps in the foreground objects of a binary image by performing dilation followed by erosion.

```python
print("Using:", os.path.basename(SELECTED_IMG_PATH))

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))

def closing(image, kernel):
    return erosion(dilation(image, kernel), kernel)

closed = closing(FG_MASK, kernel)

plt.figure(figsize=(10,4))
plt.subplot(1,3,1); plt.imshow(RGB_ORIG); plt.title("Original"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(FG_MASK, cmap="gray"); plt.title("Binary mask"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(closed, cmap="gray"); plt.title("D) Closing"); plt.axis("off")
plt.tight_layout(); plt.show()
```

(E). In this exercise, you will implement a **Morphological Gradient** function. Morphological gradient is a morphological operation that highlights the edges of objects in a binary image by calculating the difference between dilation and erosion of the image.

```python
print("Using:", os.path.basename(SELECTED_IMG_PATH))

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))

def morph_gradient(image, kernel):
    return cv2.subtract(dilation(image, kernel), erosion(image, kernel))

grad = morph_gradient(FG_MASK, kernel)

# Optional overlay in red
overlay = RGB_ORIG.copy()
edges = grad > 0
overlay[edges] = [255, 0, 0]

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.imshow(RGB_ORIG); plt.title("Original"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(FG_MASK, cmap="gray"); plt.title("Binary mask"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(grad, cmap="gray"); plt.title("E) Morphological Gradient"); plt.axis("off")
plt.tight_layout(); plt.show()

plt.figure(figsize=(4,4))
plt.imshow(overlay); plt.title("Gradient edges overlay"); plt.axis("off")
plt.show()
```

# **Watermarking** (OPTIONAL)

In this exercise, you will implement watermarking using a combination of linear filtering, non-linear filtering, and morphological operations. The objective is to embed a watermark into an image by enhancing its structure and seamlessly blending it with the original. You will use techniques such as smoothing, edge detection, and morphological operations to improve the clarity of the watermark while adjusting its visibility through alpha blending.


**Steps**:

1. Pre-process the watermark using filters like Gaussian and/or Mean to smooth out details.

2. Enhance edges of the watermark using filters such as Laplacian and/or Sobel.

3. Refine the watermark's structure with morphological operations like dilation and/or erosion.

4. Blend the processed watermark with the original image using alpha blending via the `cv2.addWeighted()` function.

**Customization Tips**:
- Kernel sizes: Adjust kernel sizes for filters like Gaussian ((7,7)), Sobel (ksize=3), and morphological operations ((5,5)).

- Alpha blending factor: Modify the alpha value (e.g., alpha = 0.3) to control the transparency and visibility of the watermark.

- Filter combinations: Experiment by changing the order of filters and morphological operations to achieve different effects. For example, applying morphological operations earlier may lead to varied results.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

original_image = cv2.imread('lenna.png')                              
watermark = cv2.imread('watermark.png', cv2.IMREAD_GRAYSCALE)         
if original_image is None or watermark is None:
    raise FileNotFoundError("Missing 'lenna.png' or 'watermark.png'.")

wm_blur = cv2.GaussianBlur(watermark, (7, 7), 0)

grad_x = cv2.Sobel(wm_blur, cv2.CV_32F, 1, 0, ksize=3)
grad_y = cv2.Sobel(wm_blur, cv2.CV_32F, 0, 1, ksize=3)
grad_magnitude = cv2.magnitude(grad_x, grad_y)

def to_uint8(img32):
    gmin, gmax = float(img32.min()), float(img32.max())
    if gmax - gmin < 1e-6:
        return np.zeros_like(img32, dtype=np.uint8)
    return np.uint8(255 * (img32 - gmin) / (gmax - gmin))

grad_x_u8 = to_uint8(np.abs(grad_x))
grad_y_u8 = to_uint8(np.abs(grad_y))
grad_mag_u8 = to_uint8(grad_magnitude)

kernel = np.ones((5, 5), np.uint8)
wm_morph = cv2.dilate(grad_mag_u8, kernel, iterations=1)
wm_morph = cv2.erode(wm_morph, kernel, iterations=1)
wm_final_gray = cv2.GaussianBlur(wm_morph, (5, 5), 0)

h, w = original_image.shape[:2]
target_w = int(0.3 * w)
scale = target_w / wm_final_gray.shape[1]
target_h = int(wm_final_gray.shape[0] * scale)
wm_resized = cv2.resize(wm_final_gray, (target_w, target_h), interpolation=cv2.INTER_AREA)

wm_canvas = np.zeros_like(original_image, dtype=np.uint8)
y_margin, x_margin = 20, 20
y1 = max(0, h - target_h - y_margin)
x1 = max(0, w - target_w - x_margin)
y2, x2 = y1 + target_h, x1 + target_w

wm_color = cv2.cvtColor(wm_resized, cv2.COLOR_GRAY2BGR)
wm_canvas[y1:y2, x1:x2] = wm_color

alpha = 0.3
watermarked_image = cv2.addWeighted(original_image, 1.0, wm_canvas, alpha, 0.0)

def bgr2rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(bgr2rgb(original_image))
axes[0].set_title('Original Image', fontsize=6)
axes[0].axis('off')

axes[1].imshow(wm_resized, cmap='gray')
axes[1].set_title('Processed Watermark', fontsize=6)
axes[1].axis('off')

axes[2].imshow(bgr2rgb(watermarked_image))
axes[2].set_title('Watermarked Image', fontsize=6)
axes[2].axis('off')

fig.tight_layout()

out_dir = Path('outputs'); out_dir.mkdir(exist_ok=True)
fig.savefig(out_dir / 'preview_3_side_by_side.png', dpi=200, bbox_inches='tight')

plt.show()    
plt.close(fig)
```
